## Imports & Setup

```python
import numpy as np
import os
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, roc_auc_score

from sklearn.ensemble import ExtraTreesClassifier, VotingClassifier
from sklearn.linear_model import RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.calibration import CalibratedClassifierCV

from imblearn.over_sampling import SMOTE
import lightgbm as lgb

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
```

## Load Dataset

```python
# Load dataset and basic checks
csv_path = 'breast_cancer.csv'
assert os.path.exists(csv_path), 'breast_cancer.csv not found in project root.'

df = pd.read_csv(csv_path)
print('Original shape:', df.shape)
print(df.head())
print(df.info())

# Drop non-informative identifiers if present
if 'id' in df.columns:
    df = df.drop(columns=['id'])

# Encode target: diagnosis (M=1, B=0)
assert 'diagnosis' in df.columns, 'Expected target column "diagnosis" not found.'
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0}).astype(int)

# Handle missing values (median for numeric)
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
for col in num_cols:
    if df[col].isna().any():
        df[col] = df[col].fillna(df[col].median())

print('Post-clean shape:', df.shape)
df.head()

```

## Lightweight Feature Engineering (no target leakage)

```python
fe = df.copy()
eps = 1e-9

# shape ratios (robust geometric relations)
if set(["area_mean", "perimeter_mean", "radius_mean"]).issubset(fe.columns):
    fe["area_perimeter_ratio_mean"]  = fe["area_mean"]      / (fe["perimeter_mean"] + eps)
    fe["perimeter_radius_ratio_mean"]= fe["perimeter_mean"] / (fe["radius_mean"] + eps)

if set(["area_worst", "perimeter_worst", "radius_worst"]).issubset(fe.columns):
    fe["area_perimeter_ratio_worst"]   = fe["area_worst"]      / (fe["perimeter_worst"] + eps)
    fe["perimeter_radius_ratio_worst"] = fe["perimeter_worst"] / (fe["radius_worst"] + eps)

# worst/mean stability ratios
pairs = [
    ("radius_worst","radius_mean"),
    ("perimeter_worst","perimeter_mean"),
    ("area_worst","area_mean"),
    ("compactness_worst","compactness_mean"),
    ("concavity_worst","concavity_mean"),
    ("concave points_worst","concave points_mean"),
]
for w, m in pairs:
    if w in fe.columns and m in fe.columns:
        fe[f"{w}_over_{m}"] = fe[w] / (fe[m] + eps)

print("New cols:", [c for c in fe.columns if c not in df.columns])
fe.head()
```

```python
results = {}

etc = ExtraTreesClassifier(n_estimators=400, random_state=RANDOM_STATE, class_weight='balanced')
rc = RidgeClassifier(random_state=RANDOM_STATE)
lda = LinearDiscriminantAnalysis()

try:
    from lightgbm import LGBMClassifier
    lgbm = LGBMClassifier(random_state=RANDOM_STATE, class_weight='balanced')
    lgbm_available = True
except ImportError:
    lgbm = None
    lgbm_available = False

from sklearn.metrics import ConfusionMatrixDisplay

def evaluate(name, model, Xtr, ytr, Xte, yte):
    model.fit(Xtr, ytr)
    y_pred = model.predict(Xte)
    acc = accuracy_score(yte, y_pred)
    print(f'{name} Accuracy: {acc:.4f}')
    print(classification_report(yte, y_pred, target_names=['Benign(0)', 'Malignant(1)']))
    cm = confusion_matrix(yte, y_pred)
    ConfusionMatrixDisplay(cm, display_labels=['Benign', 'Malignant']).plot(cmap='Blues')
    plt.title(f'{name} - Confusion Matrix')
    plt.show()
    return acc

def evaluate_and_store(name, model, Xtr, ytr, Xte, yte, results_dict):
    print('='*80)
    acc = evaluate(name, model, Xtr, ytr, Xte, yte)
    results_dict[name] = acc

evaluate_and_store('Extra Trees', etc, X_train_bal, y_train_bal, X_test_scaled, y_test, results)
evaluate_and_store('Ridge Classifier', rc, X_train_bal, y_train_bal, X_test_scaled, y_test, results)
evaluate_and_store('LDA', lda, X_train_bal, y_train_bal, X_test_scaled, y_test, results)

if lgbm_available:
    evaluate_and_store('LightGBM', lgbm, X_train_bal, y_train_bal, X_test_scaled, y_test, results)
else:
    print('Skipping LightGBM (not installed).')

results
```

## Tuned Soft-Voting Ensemble (ETC + Calibrated Ridge + LGBM + LDA)

```python
## Tuned Soft-Voting Ensemble (ETC + Calibrated Ridge + LGBM + LDA)

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import roc_curve, auc
import time

# Reuse train/test built earlier: X_train_res_scaled, y_train_res, X_test_scaled, y_test

# Base learners
etc_tuned = ExtraTreesClassifier(
    n_estimators=800,
    max_features='sqrt',
    min_samples_split=2,
    min_samples_leaf=1,
    bootstrap=False,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

ridge_base = RidgeClassifier(random_state=RANDOM_STATE)
rc_cal_tuned = CalibratedClassifierCV(estimator=ridge_base, method='sigmoid', cv=5)

lda_tuned = LinearDiscriminantAnalysis()

lgbm_tuned = lgb.LGBMClassifier(
    n_estimators=700,
    learning_rate=0.05,
    num_leaves=31,
    subsample=0.9,
    colsample_bytree=0.9,
    reg_lambda=0.0,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

soft_voting = VotingClassifier(
    estimators=[('etc', etc_tuned), ('rc', rc_cal_tuned), ('lgbm', lgbm_tuned), ('lda', lda_tuned)],
    voting='soft',
    weights=[2, 1, 2, 1],
    n_jobs=-1
)

# Fit ensemble and evaluate base accuracy
start_train = time.time()
soft_voting.fit(X_train_res_scaled, y_train_res)
train_time = time.time() - start_train

start_pred = time.time()
y_proba = soft_voting.predict_proba(X_test_scaled)[:, 1]
y_pred_default = (y_proba >= 0.5).astype(int)
pred_time = time.time() - start_pred

acc_default = accuracy_score(y_test, y_pred_default)
print(f"Soft Voting (default 0.5) Accuracy: {acc_default:.4f}")
print(f"Training time: {train_time:.4f}s, Prediction time: {pred_time:.4f}s")
print(classification_report(y_test, y_pred_default, target_names=['Benign(0)', 'Malignant(1)']))
cm = confusion_matrix(y_test, y_pred_default)
ConfusionMatrixDisplay(cm, display_labels=['Benign', 'Malignant']).plot(cmap='Blues')
plt.title('Soft Voting (default 0.5) - Confusion Matrix')
plt.show()

# Threshold sweep to target >= 0.99 accuracy, preferring higher recall for class 1 ties
thresholds = np.linspace(0.2, 0.8, 121)
best_acc = -1.0
best_thr = 0.5
best_metrics = None
for thr in thresholds:
    yp = (y_proba >= thr).astype(int)
    acc = accuracy_score(y_test, yp)
    if acc > best_acc or (np.isclose(acc, best_acc) and thr<best_thr):
        best_acc = acc
        best_thr = thr
        best_metrics = (classification_report(y_test, yp, output_dict=True), confusion_matrix(y_test, yp))

print(f"Best threshold: {best_thr:.3f} with accuracy {best_acc:.4f}")

# Plot ROC and selected operating point
fpr, tpr, roc_thr = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)
plt.figure()
plt.plot(fpr, tpr, label=f'ROC AUC = {roc_auc:.3f}')
# Mark chosen threshold
# Find closest threshold index
idx_thr = np.argmin(np.abs(roc_thr - best_thr)) if len(roc_thr)>0 else None
if idx_thr is not None and idx_thr < len(fpr):
    plt.scatter([fpr[idx_thr]], [tpr[idx_thr]], c='red', label=f'Th={best_thr:.2f}')
plt.plot([0,1],[0,1],'k--')
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC Curve - Soft Voting')
plt.legend()
plt.tight_layout()
plt.show()

# Report best-threshold confusion matrix
rep_dict, cm_best = best_metrics
print(classification_report(y_test, (y_proba>=best_thr).astype(int), target_names=['Benign(0)','Malignant(1)']))
ConfusionMatrixDisplay(cm_best, display_labels=['Benign','Malignant']).plot(cmap='Blues')
plt.title(f'Soft Voting @ {best_thr:.2f} - Confusion Matrix')
plt.show()

soft_voting_best = soft_voting
best_threshold = best_thr
soft_voting_acc = best_acc
```

## Complexity Analysis (Training and Prediction Time)

```python
## Complexity Analysis (Training and Prediction Time)

import pandas as pd
import time

records = []

def time_model(name, model, Xtr, ytr, Xte, reps_pred=5):
    t0 = time.time()
    model.fit(Xtr, ytr)
    t_train = time.time() - t0
    # average multiple prediction passes for stability
    t0 = time.time()
    for _ in range(reps_pred):
        _ = model.predict_proba(Xte) if hasattr(model, 'predict_proba') else model.decision_function(Xte)
    t_pred = (time.time() - t0) / reps_pred
    return t_train, t_pred

candidates = {
    'Extra Trees': etc_tuned,
    'Calibrated Ridge': rc_cal_tuned,
    'LightGBM': lgbm_tuned,
    'LDA': lda_tuned,
    'Ensemble (Soft Voting)': soft_voting_best,
}

for name, mdl in candidates.items():
    # clone a fresh instance when necessary (avoid reusing fitted state)
    from sklearn.base import clone
    m = clone(mdl)
    tr, pr = time_model(name, m, X_train_res_scaled, y_train_res, X_test_scaled)
    records.append({'Model': name, 'Train Time (s)': tr, 'Pred Time per pass (s)': pr})

complexity_df = pd.DataFrame(records).sort_values('Train Time (s)')
complexity_df.reset_index(drop=True, inplace=True)
complexity_df
```
